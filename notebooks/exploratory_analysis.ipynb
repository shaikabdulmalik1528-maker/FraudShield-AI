# FraudShield AI — Exploratory Data Analysis (EDA)
### Dataset: European Credit Card Fraud Detection (284,807 transactions)

This notebook conducts structured exploratory analysis to investigate class imbalance, feature distributions, outlier patterns, and correlation structure.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set dark fintech theme for plotting
plt.style.use('dark_background')
data_path = Path('../data/raw/creditcard.csv')
df = pd.read_csv(data_path)
print(f"Dataset shape: {df.shape}")
```

```python
# Check missing values and duplicates
print(f"Missing values: {df.isnull().sum().sum()}")
print(f"Duplicates: {df.duplicated().sum()}")
class_counts = df['Class'].value_counts()
print(f"Legitimate (0): {class_counts[0]:,} | Fraudulent (1): {class_counts[1]:,}")
print(f"Fraud Percentage: {class_counts[1] / len(df) * 100:.4f}%")
```

```python
# Class Imbalance Visualization
fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(['Legitimate (0)', 'Fraudulent (1)'], [class_counts[0], class_counts[1]], color=['#10B981', '#F43F5E'])
ax.set_yscale('log')
ax.set_title('Transaction Class Imbalance (Log Scale)')
plt.show()
```

```python
# Correlation of PCA features with Target Class
corrs = df.corr()['Class'].drop('Class').sort_values()
fig, ax = plt.subplots(figsize=(10, 6))
colors = ['#F43F5E' if v > 0 else '#06B6D4' for v in corrs.values]
ax.barh(corrs.index, corrs.values, color=colors)
ax.set_title('Feature Correlation with Fraud (Class)')
plt.show()
```
